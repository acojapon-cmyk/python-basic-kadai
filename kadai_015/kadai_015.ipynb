```
class Human:
    def __init__(self, name="", age=""):   # 初期化時にnameとageを受け取る
        self.name = name
        self.age = age

    # メソッドを定義する
    def set_name(self, name):
        self.name = name

    def set_age(self, age):
        self.age = age

    def show_name(self):
        print("名前:", self.name)

    def show_age(self):
        print("年齢:", self.age)

    # 名前と年齢を出力するメソッド
    def printinfo(self):
        print(f"名前: {self.name}, 年齢: {self.age}")


# インスタンス化する（初期値を設定）
human = Human("権平", 25)

# メソッドにアクセスして実行する
human.show_name()
human.show_age()

# printinfo メソッドを使用して名前と年齢を出力
human.printinfo()
```
