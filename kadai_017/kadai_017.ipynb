```
class Human:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def printinfo(self):
        print(f"名前: {self.name}, 年齢: {self.age}")

    def check_adult(self):
        if self.age >= 20:
            print(f"{self.name}は成人です。")
        else:
            print(f"{self.name}は成人ではありません。")

# Humanクラスのインスタンスを作成
humanA = Human("孝雄", 22)
humanB = Human("孝仁", 17)
humanC = Human("悠仁", 55)
humanD = Human("芳江", 13)

# Humanクラスのインスタンスをリストに追加
human_instances = [humanA, humanB, humanC, humanD]

# リストに対して繰り返し処理を使い、各インスタンスのcheck_adultメソッドを呼び出す
print("--- 各インスタンスの大人判定 ---")
for human_instance in human_instances:
    human_instance.check_adult()


```
