```python
price1 = 100
price2 = 200

def total_with_tax(p1, p2, tax_rate): # price1, price2, taxを引数として渡す
    subtotal = p1 + p2
    total_with_tax = subtotal * tax_rate
    return total_with_tax

# 税率を定義
tax_rate = 1.1

# 関数を呼び出し、結果を変数に格納
final_amount = total_with_tax(price1, price2, tax_rate)

# 結果を出力
print(final_amount) # 出力
```
