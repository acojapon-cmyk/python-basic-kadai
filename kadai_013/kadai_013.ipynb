```python
# 与えられた引数priceに消費税を加算し、その値を出力する関数を定義する
def calculate_total_with_tax(price, tax):
    total = price + tax * price / 100
    return total # 計算結果を戻り値として返す

# 関数を呼び出し、引数として価格と税率を渡す
# 戻り値を変数に代入して利用する
result = calculate_total_with_tax(110, 10) # 例として110円, 10%を渡す
print(f"{result}円") # 戻り値を使って出力
```
